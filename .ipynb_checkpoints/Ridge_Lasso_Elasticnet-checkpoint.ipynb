```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures,OneHotEncoder
from sklearn.metrics import root_mean_squared_error,r2_score
import seaborn as sns
os.chdir("E:/Machine_Learning/Datasets") 
```

```python
df = pd.read_csv('Housing.csv')
df.head()
```

```python
df.dropna(inplace=True)
X , y = df.drop('price',axis=1) , df['price']
categorical = list(X.columns[X.dtypes == 'str'])
numerical = list(X.columns[X.dtypes == 'int'])
ohe = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas') # Drop first column Ex.('no' is dropped) and 
X_ohe = ohe.fit_transform(X[categorical])
```

```python
X_enc = pd.concat([X_ohe,X[numerical]],axis=1)
X_enc
```

## Linear Regression

```python
X_train,X_test,y_train,y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)

lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
rmse = root_mean_squared_error(y_pred,y_test)
rmse
```

```python
lr.coef_
```

## Ridge Regression

```python
X_train,X_test,y_train,y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)

ridge = Ridge(alpha=1)
ridge.fit(X_train,y_train)
y_pred = ridge.predict(X_test)
rmse = root_mean_squared_error(y_pred,y_test)
rmse
```

```python
ridge.coef_
```

```python
alphas = [0.1,0.25,0.7,1.5,2,5,10,50]
for i in alphas:
    ridge = Ridge(alpha=i)
    ridge.fit(X_train,y_train)
    y_pred = ridge.predict(X_test)
    rmse = root_mean_squared_error(y_pred,y_test)
    print(f'alpha = {i} | rmse : {rmse}')
```

# Lasso Regression

```python
X_train,X_test,y_train,y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)

ridge = Lasso(alpha=1)
ridge.fit(X_train,y_train)
y_pred = ridge.predict(X_test)
rmse = root_mean_squared_error(y_pred,y_test)
rmse
```

```python
alphas = [0.1,0.25,0.7,1.5,2,5,10,50]
for i in alphas:
    ridge = Lasso(alpha=i)
    ridge.fit(X_train,y_train)
    y_pred = ridge.predict(X_test)
    rmse = root_mean_squared_error(y_pred,y_test)
    print(f'alpha = {i} | rmse : {rmse}')
```

# ElasticNet 

```python
from sklearn.linear_model import ElasticNet
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)

ridge = ElasticNet(alpha=1)
ridge.fit(X_train,y_train)
y_pred = ridge.predict(X_test)
rmse = root_mean_squared_error(y_pred,y_test)
rmse
```

```python
alphas = [0.1,0.25,0.7,1.5,2,5,10,50]
ratios = [0.1,0.25,0.5,0.75,1]
scores = []
for i in alphas:
    for r in ratios:
        el = ElasticNet(alpha=i,l1_ratio=r)
        el.fit(X_train,y_train)
        y_pred = el.predict(X_test)
        scores.append([i,r,root_mean_squared_error(y_test,y_pred)])

df_scores = pd.DataFrame(scores,columns=['alpha','l1','rmse'])
df_scores
```
