```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures,OneHotEncoder
from sklearn.metrics import root_mean_squared_error,r2_score
import seaborn as sns
os.chdir("E:/Machine_Learning/Datasets") 
```

```python
salary = pd.read_csv('Exp_Salaries.csv')
salary.head()
```

```python
X ,y=salary.drop('Salary',axis=1) , salary['Salary']
```

```python
categorical = list(X.columns[X.dtypes == 'str'])
numerical = list(X.columns[X.dtypes == 'int'])
```

```python
ohe = OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
X_ohe = ohe.fit_transform(X[categorical])
```

```python
X_final = pd.concat([X[numerical],X_ohe],axis=1)
X_final.head()
```

```python
X_train,X_test,y_train,y_test =train_test_split(X_final,y,test_size=0.3,random_state=26)
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
r2 = r2_score(y_test,y_pred)
print('r2 : ',r2)
```

# Ridge

```python
alphas = np.linspace(0.001,10,20)
scores = []
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_train,y_train)
    y_pred = ridge.predict(X_test)
    r2 = r2_score(y_test,y_pred)
    scores.append([a,r2])
```

```python
df_r2_scores_ridge = pd.DataFrame(scores,columns=['alpha','r2 score'])
df_r2_scores_ridge.nlargest(1,'r2 score')
```

# Lasso

```python
alphas = np.linspace(0.001,10,20)
scores = []
for a in alphas:
    lasso = Lasso(alpha=a)
    lasso.fit(X_train,y_train)
    y_pred = lasso.predict(X_test)
    r2 = r2_score(y_test,y_pred)
    scores.append([a,r2])

df_r2_scores_lasso = pd.DataFrame(scores,columns=['alpha','r2 score'])
df_r2_scores_lasso.nlargest(1,'r2 score')
```

# ElasticNet

```python
alphas = np.linspace(0.001,10,20)
ratios = np.linspace(0.001,1,20)
scores = []
for a in alphas:
    for r in ratios:
        el = ElasticNet(alpha=a,l1_ratio=r)
        el.fit(X_train,y_train)
        y_pred = el.predict(X_test)
        r2 = r2_score(y_test,y_pred)
        scores.append([a,r,r2])

df_r2_scores_elastic = pd.DataFrame(scores,columns=['alpha','l1 ratio','r2 score'])
df_r2_scores_elastic.nlargest(1,'r2 score')
```
