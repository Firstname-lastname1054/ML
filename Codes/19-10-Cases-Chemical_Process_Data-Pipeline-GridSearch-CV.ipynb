```python
import numpy as np
import pandas as pd

from sklearn.model_selection import GridSearchCV,KFold
from sklearn.metrics import roc_auc_score , log_loss


from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsRegressor
import os

import warnings
from sklearn.exceptions import ConvergenceWarning
warnings.filterwarnings("ignore",category = ConvergenceWarning)
os.chdir('E:/Machine_Learning/Cases/Chemical_Process_Data/')
```

```python
chem = pd.read_csv('ChemicalProcess.csv')
chem.head()
```

```python
X,y = chem.drop('Yield',axis=1),chem['Yield']
std_scaler = StandardScaler()
mm_scaler = MinMaxScaler()
imputer = SimpleImputer()
knn = KNeighborsRegressor()
kfold = KFold(n_splits=5,shuffle=True,random_state=26)
pipe = Pipeline([('IMP',imputer),('SCL',std_scaler),('KNN',knn)])
pipe.get_params()
```

```python
params = {
    'IMP__strategy' : ['mean','median'],
    'SCL' : [std_scaler,mm_scaler,None],
    'KNN__n_neighbors' : np.arange(1,11)
}
```

```python
gcv = GridSearchCV(pipe,param_grid=params,cv = kfold,n_jobs = -1)
gcv.fit(X,y)
print('Best Param : ',gcv.best_params_)
print('Best Score : ',gcv.best_score_)
```
