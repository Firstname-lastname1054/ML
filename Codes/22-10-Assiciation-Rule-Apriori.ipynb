```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
import os
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
fp_df = pd.read_csv("Faceplate.csv",index_col=0)
fp_df
```

```python
fp_df = pd.read_csv("Faceplate.csv",index_col=0)
fp_df = fp_df.astype(bool)
itemsets = apriori(fp_df,min_support=0.5,use_colnames=True)
itemsets
```

```python
fp_df = fp_df.astype(bool)
itemsets = apriori(fp_df,min_support=0.2,use_colnames=True)
itemsets
```

```python
rules =association_rules(itemsets,metric='confidence',min_threshold=0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by=['confidence','lift','support'],ascending = False)
```

## relevant rules will be those rules which have lift value greater than 1 so we filter that

```python
relevant_rules = rules[rules['lift']>1]
relevant_rules
```
