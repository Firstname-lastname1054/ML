```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import os
os.chdir('E:/Machine_Learning/Cases/Wisconsin')
```

```python
bcancer = pd.read_csv('BreastCancer.csv',index_col=0)
bcancer.head()
```

```python
X,y = bcancer.drop('Class',axis=1) , bcancer['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train,y_train)
y_pred= knn.predict(X_test)
acc_score = accuracy_score(y_test,y_pred)
acc_score
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train,y_train)
y_pred= knn.predict(X_test)
acc_score = accuracy_score(y_test,y_pred)
acc_score
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
y_pred= knn.predict(X_test)
acc_score = accuracy_score(y_test,y_pred)
acc_score
```

```python
ks = np.arange(1,16)
scores = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train,y_train)
    y_pred = knn.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    scores.append([k,acc])
df_scores = pd.DataFrame(scores,columns=['k','Accuracy Score'])
df_scores
```

```python
df_scores.sort_values(by='Accuracy Score',ascending=False)
```

```python
df_scores.nlargest(1,'Accuracy Score')
```
