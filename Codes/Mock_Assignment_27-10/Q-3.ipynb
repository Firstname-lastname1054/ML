```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold,GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('E:/Machine_Learning/Cases/Vehicle Silhouettes/Vehicle.csv')
df.head()
```

```python
X,y = df.drop('Class',axis=1),df["Class"]
```

```python
kfold = StratifiedKFold(n_splits=5,shuffle=True,random_state=26)
dtc = DecisionTreeClassifier(random_state=26)
rf = RandomForestClassifier(random_state=26)
lr = LogisticRegression()
```

```python

params = {
    'max_depth':[2,4,6],
    'min_samples_split':[2,3,5],
    'min_samples_leaf':[1,3,5]
    
}
gcv = GridSearchCV(dtc,params,cv=kfold,scoring='neg_log_loss',n_jobs=-1)
gcv.fit(X,y)
print("Best param : ", gcv.best_params_)
print("Best score : ", gcv.best_score_)
```

```python
params = {
    'max_features':[2,3,4,5]
}
gcv = GridSearchCV(rf,params,cv=kfold,scoring = 'neg_log_loss',n_jobs=-1)
gcv.fit(X,y)
print("Best param : ", gcv.best_params_)
print("Best score : ", gcv.best_score_)
```

```python
params = {
    'solver':['lbfgs','newton-cg','newton-cholesky','sag','saga']
}

gcv = GridSearchCV(lr,params,cv=kfold,scoring = 'neg_log_loss',n_jobs=-1)
gcv.fit(X,y)
print("Best param : ", gcv.best_params_)
print("Best score : ", gcv.best_score_)
```
