```python
import os 
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE
os.chdir("E:/Machine_Learning/Cases/Wisconsin/")
```

```python
df = pd.read_csv('BreastCancer.csv',index_col=0)
df.head()
```

```python
X,y = df.drop('Class',axis = 1),df['Class']
tsne = TSNE(random_state=26).set_output(transform='pandas')
t_comps = tsne.fit_transform(X,y)
```

```python
t_comps.head()
```

```python
t_comps['class']=y
```

```python

# sns.scatterplot(data = t_comps,x = t_comps['tsne0'],y = t_comps['tsne1'],hue=t_comps['class'])
sns.scatterplot(data = t_comps,x = 'tsne0',y ='tsne1',hue='class')
plt.show()
```
