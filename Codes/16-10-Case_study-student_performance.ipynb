```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
import os
os.chdir('E:/Machine_Learning/Cases/student_performance')
```

```python
df = pd.read_csv('student_dataset_10000_rows.csv')
df.head()
```

```python
X,y = df.drop('placement_status',axis=1),df['placement_status']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
scaler = StandardScaler()
```

```python
ks = np.arange(1,16)
scores = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    pipe1 = Pipeline([('SCL',scaler),('KNN',knn)])
    pipe1.fit(X_train,y_train)
    y_pred = pipe1.predict(X_test)
    scores.append([k,accuracy_score(y_test,y_pred)])

df_scores = pd.DataFrame(scores,columns=['K','Accuracy'])
df_scores.sort_values(by=['Accuracy'],ascending=False)
```
