### Using fit() and transform() and fit_transform()

```python
housing  = pd.read_csv('Housing.csv')
X , y = housing.drop('price',axis = 1) , housing['price']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
ohe = OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')

ct = make_column_transformer((ohe,make_column_selector(dtype_include='object')),
                            ('passthrough',make_column_selector(dtype_include=['int64','float64'])),
                            verbose_feature_names_out=False)

ct = ct.set_output(transform='pandas')

X_train_transformed = ct.fit_transform(X_train)
```

```python
X_test_transformed = ct.transform(X_test) # use only transform() on testing data
```

```python
lr = LinearRegression()
lr.fit(X_train_transformed,y_train)
y_pred = lr.predict(X_test_transformed)
```

```python
r2_score(y_test,y_pred)
```
