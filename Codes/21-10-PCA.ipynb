```python
import os 
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
df = pd.read_csv('milk.csv',index_col='Animal')
df.head()
```

```python
scaler = StandardScaler().set_output(transform = 'pandas')
scl_milk = scaler.fit_transform(df)
scl_milk.head()
```

```python
prcomp = PCA().set_output(transform='pandas')
pca_data = prcomp.fit_transform(scl_milk)
pca_data.head()
```

```python
df.shape,pca_data.shape
```

```python
pca_data.var()
```

```python
sigma = scl_milk.cov()
values, vectors = np.linalg.eig(sigma)
print('Eigen Values : ' , values)
```

```python
print('Eigen Values : ' , vectors)
```
