```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold,GridSearchCV
from tqdm import tqdm
from sklearn.svm import SVC
import os
os.chdir('E:/Machine_Learning/Cases/Vehicle Silhouettes')
```

```python
vehicle= pd.read_csv("Vehicle.csv")
vehicle.head()
```

```python
X ,y = vehicle.drop('Class',axis=1),vehicle['Class']
```

```python
folds = StratifiedKFold(random_state=26,shuffle=True)
scaler = StandardScaler()
pca = PCA()
lr = LogisticRegression()
pipe = Pipeline([('SCL',scaler),('PCA',pca),("LR",lr)])
pipe.get_params()
```

```python
params = {
    'PCA__n_components':np.arange(2,19),  # here we are specifying how many principal components should we take instead of that we can use floating numbers which will set threshold 
    'LR__C': np.linspace(0.001,5,10)
}
gcv = GridSearchCV(pipe,param_grid=params,scoring='neg_log_loss',cv=folds,n_jobs=-1)
gcv.fit(X,y)
```

```python
print("Best param : ", gcv.best_params_)
print("Best score : ", gcv.best_score_)
```

### Using SVM instead of LR

```python
folds = StratifiedKFold(shuffle = True ,random_state=26)
scale = StandardScaler()
svc = SVC(probability=True)
pca = PCA()
pipe = Pipeline([('SCA',scale),('PCA',pca),('SVC',svc)])
pipe.get_params()
```

```python
params = {'PCA__n_components' : np.arange(1,19),
         'SVC__C' : np.linspace(0.001,10,20)}
grid = GridSearchCV(pipe,cv=folds,n_jobs = -1,param_grid=params,scoring='neg_log_loss')
grid.fit(X,y)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```
