```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier,RandomForestRegressor
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
import matplotlib.pyplot as plt
import os
os.chdir('E:/Machine_Learning/Cases/Wisconsin/')
```

```python
wisc = pd.read_csv("BreastCancer.csv",index_col=0)
wisc.head()
```

```python
X,y = wisc.drop('Class',axis = 1),wisc['Class']
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
rf = RandomForestClassifier(n_estimators=25,max_features=3)
rf.fit(X_train,y_train)
y_pred = rf.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
m_feat=[2,3,4,5]
samples = [25,50,100]
scores = []
for m in m_feat:
    for s in samples:
        rf = RandomForestClassifier(n_estimators=s,max_features=m,random_state=26)
        rf.fit(X_train,y_train)
        y_pred = rf.predict(X_test)
        scores.append([s,m,precision_score(y_test,y_pred,pos_label='Malignant')])
df_s =pd.DataFrame(scores,columns=['n','Max_features','precision'])
df_s.sort_values(by= 'precision',ascending=False)
```

```python
m_feat=[2,3,4,5]
samples = [25,50,100]
scores = []
for m in m_feat:
    for s in samples:
        rf = RandomForestClassifier(n_estimators=s,max_features=m,random_state=26)
        rf.fit(X_train,y_train)
        y_pred = rf.predict(X_test)
        scores.append([s,m,recall_score(y_test,y_pred,pos_label='Malignant'),
                       precision_score(y_test,y_pred,pos_label='Malignant'),
                       f1_score(y_test,y_pred,pos_label='Malignant')
                      ])
df_s =pd.DataFrame(scores,columns=['n','Max_features','recall','precision','f1'])
df_s
```

```python
rf = RandomForestClassifier(n_estimators=25,max_features=3)
rf.fit(X_train,y_train)
y_pred = rf.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

## Vehicle dataset

```python
os.chdir('E:/Machine_Learning/Cases/Vehicle Silhouettes/')
vehicle= pd.read_csv("Vehicle.csv")
vehicle.head()
```

```python
X,y = vehicle.drop('Class',axis=1),vehicle['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
rf = RandomForestClassifier(n_estimators=25,max_features=3)
rf.fit(X_train,y_train)
y_pred = rf.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
m_feat=[2,3,4,5]
samples = [25,50,100]
scores = []
for m in m_feat:
    for s in samples:
        rf = RandomForestClassifier(n_estimators=s,max_features=m,random_state=26)
        rf.fit(X_train,y_train)
        y_pred = rf.predict(X_test)
        scores.append([s,m,recall_score(y_test,y_pred,pos_label='van',average='weighted'),
                       precision_score(y_test,y_pred,pos_label='van',average='weighted'),
                       f1_score(y_test,y_pred,pos_label='van',average='weighted')
                      ])
df_s =pd.DataFrame(scores,columns=['n','Max_features','recall','precision','f1'])
df_s
```
