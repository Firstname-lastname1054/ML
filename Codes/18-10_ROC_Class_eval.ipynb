```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve,roc_auc_score,RocCurveDisplay
import os 
os.chdir('E:/Machine_Learning/Datasets')
```

```python
df = pd.read_csv('comp_prob.csv')
df
```

```python
y_test,y_prob1,y_prob2 = df['y_test'],df['yprob_1'],df['yprob_2']
```

### Model 1 

```python
fpr,tpr,thres = roc_curve(y_test,y_prob1)
plt.scatter(fpr,tpr,c='red')
plt.plot(fpr,tpr)
plt.text(0.4,0.6,"AUC = %.4f" % roc_auc_score(y_test,y_prob1))
plt.xlabel('1-Specifity')
plt.ylabel('Sensitivity')
plt.show()
```

### Model 2

```python
fpr,tpr,thres = roc_curve(y_test,y_prob2)
plt.scatter(fpr,tpr,c='red')
plt.plot(fpr,tpr)
plt.text(0.4,0.6,"AUC = %.4f" % roc_auc_score(y_test,y_prob2))
plt.xlabel('1-Specifity')
plt.ylabel('Sensitivity')
plt.show()
```
