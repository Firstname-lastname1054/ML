```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
os.chdir('E:/Machine_Learning/Datasets/')
```

```python
df = pd.read_csv('monthly-milk-production-pounds-p.csv',index_col=0)
df.head()
```

```python
df.index.dtype
```

```python
df.index = pd.to_datetime(df.index).to_period('M')
df.index.dtype
```

----


----

```python
df.plot()
plt.title('Monthly Milk Production')
plt.xlabel("Months")
plt.show()
```

```python
downsample = df.resample('Q').sum()
downsample.index.rename('Quarter',inplace=True)
downsample
```

----


----

```python
downsample.plot()
plt.title('Quarterly Milk Production')
plt.xlabel("Quarterly")
plt.show()
```

# Electric Production dataset
-----

```python
electric = pd.read_csv('Electric_Production.csv',index_col=0)
electric.head()
```

```python
electric.index = pd.to_datetime(electric.index).to_period('M')

electric.plot()
plt.title('Monthly Electricity Production')
plt.xlabel("Months")
plt.show()
```

```python
downsample = electric.resample('Q').sum()
downsample.index.rename('Quarter',inplace=True)
```

```python
downsample.plot()
plt.title('Quarterly Electricity Production')
plt.xlabel("Quarterly")
plt.show()
```

## Logarithmic Graph

```python
electric['IPG2211A2N'] = np.log(electric['IPG2211A2N'])
electric.plot()
plt.title('Log scaled Electricity Production')
plt.xlabel("Months")
plt.show()
```

### Logarithmic Scaling will not change trend or any pattern in data it just changes the scale
