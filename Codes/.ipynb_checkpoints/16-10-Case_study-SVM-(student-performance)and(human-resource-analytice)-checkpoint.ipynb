```python
df = pd.read_csv('HR_comma_sep.csv')
df.head()
```

```python
X , y = df.drop('left',axis =1), df['left']
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)

ohe = OneHotEncoder(sparse_output= True,drop='first')
ct = make_column_transformer((ohe,make_column_selector(dtype_include='object')),
                            ('passthrough',make_column_selector(dtype_include=['int64','float64'])),
                            verbose_feature_names_out=False)
scaler = StandardScaler()
```

```python
svm = SVC(kernel='linear')
pipe = Pipeline([('CT',ct),('SCL',scaler),('SVM',svm)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
Cs = np.linspace(0.001,10,20)
scores = []
for c in tqdm(Cs):
    svm = SVC(kernel='linear',C=c)
    pipe = Pipeline([('CT',ct),('SCL',scaler),('SVM',svm)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([c,accuracy_score(y_test,y_pred)])

df_score = pd.DataFrame(scores,columns=['C','Accuracy'])
df_score.sort_values(by = 'Accuracy',ascending=False)
```

# what if data looks like this ?


#### This data can not be seperated by a straight line so, 
## We use poly

we just select kernel as 'poly'

```python
svm = SVC(kernel='poly') # here default degree is 2
pipe = Pipeline([('CT',ct),('SCL',scaler),('SVM',svm)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
Cs = np.linspace(0.001,10,20)
scores = []
for c in tqdm(Cs):
    for d in range(2,5):
        svm = SVC(kernel='poly',C=c,degree=d)
        pipe = Pipeline([('CT',ct),('SCL',scaler),('SVM',svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,d,accuracy_score(y_test,y_pred)])

df_score = pd.DataFrame(scores,columns=['C','Degree','Accuracy'])
df_score.sort_values(by = 'Accuracy',ascending=False)
```
