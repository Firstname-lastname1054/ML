```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,MinMaxScaler,OneHotEncoder
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
from tqdm import tqdm
import os
os.chdir('E:/Machine_Learning/Cases/Vehicle Silhouettes/')
```

```python
vehicle= pd.read_csv("Vehicle.csv")
vehicle.head()
```

```python
X,y = vehicle.drop('Class',axis=1),vehicle['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

## GradientBoosting Classifier

```python
learn_rates = np.linspace(0.01,0.9,10)
max_depths = [None,2,3,4]
n_estimators = [25,50,100]
scores = []
for l in tqdm(learn_rates):
    for m in max_depths:
        for n in n_estimators:
            gbm = GradientBoostingClassifier(random_state=26,learning_rate=l,max_depth=m,n_estimators=n)
            gbm.fit(X_train,y_train)
            y_pred = gbm.predict(X_test)
            scores.append([l,m,n,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['Learning_rate','Max_depth','n_estimators','accuracy'])
df_scores.sort_values(by='accuracy',ascending=False)
```

## XGBoost Classifier

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test) 

learn_rates = np.linspace(0.01,0.9,10)
max_depths = [None,2,3,4]
n_estimators = [25,50,100]
scores = []
for l in tqdm(learn_rates):
  for m in max_depths:
    for n in n_estimators:
      xgb = XGBClassifier(random_state=26, learning_rate=l, max_depth=m, n_estimators=n)
      xgb.fit(X_train, y_train_encoded)  
      y_pred = xgb.predict(X_test)
      scores.append([l, m, n, accuracy_score(y_test_encoded, y_pred)])
df_scores = pd.DataFrame(scores, columns=['Learning_rate', 'Max_depth', 'n_estimators', 'accuracy'])
df_scores.sort_values(by='accuracy',ascending=False)
```

## LightGBM

```python
learn_rates = np.linspace(0.01,0.9,10)
max_depths = [None,2,3,4]
n_estimators = [25,50,100]
scores = []
for l in tqdm(learn_rates):
    for m in max_depths:
        for n in n_estimators:
            lgbm = LGBMClassifier(random_state=26,learning_rate=l,max_depth=m,n_estimators=n)
            lgbm.fit(X_train,y_train)
            y_pred = lgbm.predict(X_test)
            scores.append([l,m,n,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['Learning_rate','Max_depth','n_estimators','accuracy'])
df_scores.sort_values(by='accuracy',ascending=False)
```

## CatBoosting

```python
learn_rates = np.linspace(0.01,0.9,10)
max_depths = [None,2,3,4]
n_estimators = [25,50,100]
scores = []
for l in tqdm(learn_rates):
    for m in max_depths:
        for n in n_estimators:
            cat = CatBoostClassifier(random_state=26,learning_rate=l,verbose=False,max_depth=m,n_estimators=n)
            cat.fit(X_train,y_train)
            y_pred = cat.predict(X_test)
            scores.append([l,m,n,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['Learning_rate','Max_depth','n_estimators','accuracy'])
df_scores.sort_values(by='accuracy',ascending=False)
```
