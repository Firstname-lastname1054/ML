```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
import os
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
df = pd.read_csv('Cosmetics.csv',index_col=0)
df
```

```python
fp_df = df.astype(bool)
itemset = apriori(fp_df,min_support=0.2,use_colnames=True)
itemset
```

```python
rules = association_rules(itemset,metric='confidence',min_threshold=0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by=['confidence','lift','support'],ascending = False)
```

```python
relevant_rules = rules[rules['lift']>1]
relevant_rules.sort_values(by = ['confidence','lift','support'],ascending=False)
```

# Interactive widgets

```python
from ipywidgets import interact
from ipywidgets import widgets
```

```python
def gen_rules(min_support , min_confidence):
    itemsets = apriori(fp_df, min_support=min_support, use_colnames=True)
    rules = association_rules(itemsets , metric='confidence' , min_threshold=min_confidence)
    rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift',]]
    relv_rules= rules[rules['lift']>1]
    display(relv_rules.sort_values(by=['confidence', 'lift', 'support'], ascending=False))
```

```python
gen_rules(min_support=0.3,min_confidence=0.5)
```

```python
interact(
    gen_rules , 
    min_support = widgets.FloatSlider(
        value = 0.05 , min = 0.01, max = 1 , description = 'Min Support:'
    ),
    min_confidence = widgets.FloatSlider(
        value = 0.05 , min = 0.01, max = 1 , description = 'Min Confidence:'
    )
)
```
