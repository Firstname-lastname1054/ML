```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
import seaborn as sns
os.chdir("E:/Machine_Learning/Datasets") 
```

```python
pizza = pd.read_csv('pizza.csv')
pizza.head()
```

```python
X = pizza[['Promote']]
y = pizza['Sales']
poly = PolynomialFeatures(include_bias=False).set_output(transform = 'pandas')
# Generally it gies output as np array so we use set_output(tranfrorm = 'pandas') it will create dataframe 
# include_bias = False removes the x^0 part of polynomial transformation
X_poly = poly.fit_transform(X)
X_poly.head()
```

```python
lr = LinearRegression()
lr.fit(X_poly,y)
print('Coefficients : ',lr.coef_)
print('Intercept : ',lr.intercept_)
```

## FOR degree = 3 

```python
poly3 = PolynomialFeatures(degree=3,include_bias=False).set_output(transform = 'pandas')
X_poly3 = poly3.fit_transform(X)
lr = LinearRegression()
lr.fit(X_poly3,y)
print('Coefficients : ',lr.coef_)
print('Intercept : ',lr.intercept_)
```
