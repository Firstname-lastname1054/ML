```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
import os
os.chdir('E:/Machine_Learning/Cases/Vehicle Silhouettes')
```

```python
vehicle = pd.read_csv('Vehicle.csv')
vehicle.head()
```

```python
X,y = vehicle.drop('Class',axis=1) , vehicle['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
ks = np.arange(1,16)
scores = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train,y_train)
    y_pred = knn.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    scores.append([k,acc])
df_scores = pd.DataFrame(scores,columns=['k','Accuracy Score'])
df_scores.nlargest(1,'Accuracy Score')
```

-----
#### In this dataset there are many different values for columns some columns has values in single digit and  some has near 500-900 but this creates the problem while finding the distance,... because the lower value features are ignored when there are higher value columns 

#### for example when we are taking 'Sc.Var.maxis' this column and 'Skew.maxis' this column the smaller value doesn't nake that much difference in eucledian distance but the larger value ('Sc.Var.maxis') this will have more impact on the eucledian distance .... so thats why we have to use standard scaling

-----

### Thats why the scaling is very important in KNN 
### Using StandardScaler

```python
X , y = vehicle.drop('Class',axis=1) , vehicle['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
ks = np.arange(1,16)
scores = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled,y_train)
    y_pred = knn.predict(X_test_scaled)
    acc = accuracy_score(y_test,y_pred)
    scores.append([k,acc])
df_scores = pd.DataFrame(scores,columns=['k','Accuracy Score'])
df_scores.nlargest(1,'Accuracy Score')
```

## Pipeline

```python
scaler = StandardScaler().set_output(transform='pandas')
ks = np.arange(1,16)
scores = []
for n in ks:
    knn = KNeighborsClassifier(n_neighbors=n)
    pipe = Pipeline([('SC',scaler),('KNN',knn)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([n,accuracy_score(y_test,y_pred)])

df_scores = pd.DataFrame(scores,columns=['K','Accuracy Score'])
df_scores.nlargest(1,'Accuracy Score')
```

### Using Min-Max Scaler

```python
mx = MinMaxScaler().set_output(transform='pandas')
ks = np.arange(1,16)
scores = []
for n in ks:
    knn = KNeighborsClassifier(n_neighbors=n)
    pipe = Pipeline([('MinMax',mx),('KNN',knn)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([n,accuracy_score(y_test,y_pred)])

df_scores = pd.DataFrame(scores,columns=['K','Accuracy Score'])
df_scores.nlargest(1,'Accuracy Score')
```
