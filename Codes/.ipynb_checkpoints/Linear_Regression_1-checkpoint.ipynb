```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
import seaborn as sns
os.chdir("E:/Machine_Learning/Datasets")  
# changes the current working directory of your Python environment to the specified folder (Datasets on your E: drive).
```

```python
pizza = pd.read_csv('pizza.csv')
pizza
```

```python
X = pizza[['Promote']]
y = pizza['Sales']
lr = LinearRegression()
lr.fit(X,y)
print('Coefficient : ',lr.coef_)
print('Intercept : ',lr.intercept_)
```

Thererfore we can say that the equation of the Linear regression line is :

y = 23.506 x Promotion + 5.485

```python
promo = np.array([34,89,100,22])
sales_pred = 5.4858653632529695 + 23.50640302 * promo
sales_pred
```

#### To give the data to lr.predict() we must give DataFrame ! so we can not give array directly

```python
test = pd.DataFrame({'Promote':promo})
lr.predict(test) # to give the data to lr.predict() we must give DataFrame ! so we can not give array directly
# predict applies the equation / predict applies the model
```

# Boston

```python
boston = pd.read_csv('Boston.csv')
boston
```

```python
X , y = boston.drop('medv',axis=1) , boston['medv']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=26,test_size=0.3)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
```

```python
rmse = root_mean_squared_error(y_pred,y_test)
print('root_mean_squared_error : ',rmse)
```
