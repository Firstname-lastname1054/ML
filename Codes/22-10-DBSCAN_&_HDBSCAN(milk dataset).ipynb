```python
from scipy.cluster.hierarchy import linkage,dendrogram
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans,DBSCAN,HDBSCAN
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
import os 
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
df = pd.read_csv('milk.csv',index_col=0)
df
```

```python
scaler = StandardScaler()
df_scl= scaler.fit_transform(df)
```

```python
dbs = DBSCAN(eps = 1.5,min_samples = 2)  
dbs.fit(df_scl)
dbs.labels_  # -1 indicates outliers
```

```python
len(np.unique(dbs.labels_[dbs.labels_!=-1]))
```

```python
inliner_obs = df_scl[dbs.labels_!= -1]
inliner_lbls = dbs.labels_[dbs.labels_!= -1]
silhouette_score(inliner_obs,inliner_lbls)
```

```python
epsi = np.linspace(0.1,3,10)
min_pts = [2,3,4,5]
scores=[]
for e in epsi:
    for m in min_pts:
        dbs = DBSCAN(eps = e,min_samples=m)
        dbs.fit(df_scl)
        if len(np.unique(dbs.labels_[dbs.labels_!=-1])) >=2:
            inliner_obs = df_scl[dbs.labels_!= -1]
            inliner_lbls = dbs.labels_[dbs.labels_!= -1]
            scores.append([e,m,silhouette_score(inliner_obs,inliner_lbls)])
df_dbs = pd.DataFrame(scores,columns=['Epsilon','Min Samples','Score'])
df_dbs
```
