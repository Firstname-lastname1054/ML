```python
import os 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from pca import pca
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
df = pd.read_csv('nutrient.csv',index_col=0)
df.head()
```

```python
scaler = StandardScaler().set_output(transform='pandas')
scl_df = scaler.fit_transform(df)
```

```python
prcomp = PCA().set_output(transform='pandas')
pca_data = prcomp.fit_transform(scl_df)
```

```python
model = pca(normalize=True)
res = model.fit_transform(df,row_labels=df.index)
```

```python
model.biplot()

plt.show()
```

```python
model.biplot()
for i in range(len(df)):
    plt.text(pca_data['pca0'].iloc[i],pca_data['pca1'].iloc[i],df.index[i])
plt.show()
```
