# BaggingClassifier

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from tqdm import tqdm
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
import os
os.chdir('E:/Machine_Learning/Cases/Wisconsin/')
```

```python
wisc = pd.read_csv("BreastCancer.csv",index_col=0)
wisc.head()
```

```python
X,y = wisc.drop('Class',axis = 1),wisc['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=26,test_size=0.3,stratify=y)

lr  = LogisticRegression()
bag = BaggingClassifier(estimator=lr,n_estimators=15,random_state=26)

bag.fit(X_train,y_train)
y_pred = bag.predict(X_test)
accuracy_score(y_test,y_pred)
```

## Trying different estimators

```python
dtc = DecisionTreeClassifier()
knn = KNeighborsClassifier(n_neighbors=5)
lr = LogisticRegression()
est = [lr,knn,dtc]
n_samples = [10,25,50]
scores = []
for e in est:
    for n in n_samples:
        bag = BaggingClassifier(estimator=e,n_estimators=n,random_state=26)
        bag.fit(X_train,y_train)
        y_pred = bag.predict(X_test)
        scores.append([e,n,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['Estimator','n estimator','score'])
df_scores.sort_values('score',ascending=False)
```

# BaggingRegressor

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.metrics import r2_score
import os
os.chdir('E:/Machine_Learning/Cases/Concrete_Strength/')
```

```python
df = pd.read_csv('Concrete_Data.csv')
df.head()
```

```python
X,y=df.drop('Strength',axis=1),df['Strength']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.3)
dtr = DecisionTreeRegressor(random_state=26)
bagreg = BaggingRegressor(estimator=dtr,n_estimators=15,random_state=26)
bagreg.fit(X_train,y_train)
y_pred = bagreg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
lr = LinearRegression()
ridge = Ridge()
lasso = Lasso()
dtr = DecisionTreeRegressor(random_state=26)
estimators = [lr,ridge,lasso,dtr]
sample = [10,25,50]
scores=[]
for e in estimators:
    for s in sample:
        bag = BaggingRegressor(estimator=e,n_estimators=s,random_state=26)
        bag.fit(X_train,y_train)
        y_pred=bag.predict(X_test)
        scores.append([e,s,r2_score(y_test,y_pred)])

df_s =pd.DataFrame(scores,columns=['Estimator','n','r2'])
df_s.sort_values(by= 'r2',ascending=False)
```
