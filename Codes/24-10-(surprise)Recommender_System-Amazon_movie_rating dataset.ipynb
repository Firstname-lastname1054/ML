```python
import surprise
import pandas as pd
import numpy as np
import os
os.chdir("E:/Machine_Learning/Cases_Rec_Sys/Amazon Movie Ratings/")
```

```python
df = pd.read_csv("Amazon.csv")
df
```

```python
ratings = pd.melt(df, id_vars='user_id',var_name="item_id",value_name="rating")
ratings = ratings[ratings['rating'].notna()]
```

```python
ratings
```

```python
lowest_rating=ratings['rating'].min()
highest_rating=ratings['rating'].max()
lowest_rating,highest_rating
```

```python
reader = surprise.Reader(rating_scale = (lowest_rating,highest_rating))
data = surprise.Dataset.load_from_df(ratings,reader)
type(data)
```

## Grid search cv (KNNZscore)

```python
from surprise.model_selection import GridSearchCV
from surprise.model_selection import KFold
```

```python
param_grid = { 'k': [25,40,50,65,75,85],'user_based': [False,True]}
kFold = KFold(n_splits= 5,random_state=26,shuffle=True)
gs = GridSearchCV(surprise.KNNWithZScore,param_grid,measures=['rmse','mae'],cv=kFold)
gs.fit(data)
```

```python
print(gs.best_score['rmse'])
print(gs.best_params['rmse'])
```

```python
print(gs.best_score['mae'])
print(gs.best_params['mae'])
```

## SVD

```python
param_grid = {
    'n_epochs':[50],
    'n_factors':[100,50],
    'lr_all':np.linspace(0.001,1,5),
    'reg_all':np.linspace(0.1,0.8,5)
}
kFold = KFold(n_splits= 5,random_state=26,shuffle=True)
gs = GridSearchCV(surprise.SVD,param_grid,measures=['rmse','mae'],joblib_verbose=3,cv=kFold,n_jobs=-1)
gs.fit(data)
```

```python
print(gs.best_score['rmse'])
print(gs.best_params['rmse'])
```

```python
print(gs.best_score['mae'])
print(gs.best_params['mae'])
```
