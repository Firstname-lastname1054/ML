```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis,QuadraticDiscriminantAnalysis
import os
os.chdir('E:/Machine_Learning/Cases/Image_Segmentation')
```

```python
img = pd.read_csv('Image_Segmentation.csv')
img.head()
```

## LDA

```python
X , y = img.drop('Class',axis=1) , img['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
lda = LinearDiscriminantAnalysis()
lda.fit(X_train,y_train)
y_pred = lda.predict(X_test)
accuracy_score(y_test,y_pred)
```

## QDA

```python
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train,y_train)
y_pred = qda.predict(X_test)
accuracy_score(y_test,y_pred)
```

### In this error => 
### in some of the matrix of (variance,covariance) one of the matrix have determinant as 0 
### so its inverse can't be calculated so the formula doesn't work .. so we use reg_param=  to fine tune the model 
### refer documentation of QDA for reg_param


```python
qda = QuadraticDiscriminantAnalysis(reg_param=0.1)
qda.fit(X_train,y_train)
y_pred = qda.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
satellite = pd.read_csv('Satellite.csv',sep=';')
satellite.head()
```

```python
X ,y = satellite.drop('classes',axis=1),satellite['classes']
```

# LDA

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
lda = LinearDiscriminantAnalysis()
lda.fit(X_train,y_train)
y_pred = lda.predict(X_test)
accuracy_score(y_test,y_pred)
```

## QDA

```python
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train,y_train)
y_pred = qda.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
reg = np.linspace(0,1,10)
for r in reg:
    qda = QuadraticDiscriminantAnalysis(reg_param=r)
    qda.fit(X_train,y_train)
    y_pred = qda.predict(X_test)
    acc= accuracy_score(y_test,y_pred)
    print(r,' : ',acc)
```
