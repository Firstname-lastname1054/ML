```python
from scipy.cluster.hierarchy import linkage,dendrogram
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
import os 
os.chdir("E:/Machine_Learning/Datasets/")
```

```python
milk = pd.read_csv('milk.csv',index_col=0)
milk.head()
```

```python
clf = IsolationForest(contamination=0.05,random_state=26)
clf.fit(milk)
predictions = clf.predict(milk)
print("percentage of outliers = "+str((predictions<0).mean()*100)+"%")
ab_ind = np.where(predictions<0)
print('Outliers : ')
print(ab_ind)
```

```python
milk.index[ab_ind]
```
