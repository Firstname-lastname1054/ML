```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
import os 
os.chdir('E:/Machine_Learning/Datasets')
```

```python
eval_df= pd.read_csv('eval_data.csv')
eval_df
```

```python
cm = confusion_matrix(eval_df['y_test'],eval_df['y_pred'])
disp = ConfusionMatrixDisplay(cm)
disp.plot()
plt.show()
```

```python
print(recall_score(eval_df['y_test'],eval_df['y_pred'],pos_label=0))
print(recall_score(eval_df['y_test'],eval_df['y_pred'],pos_label=1))
print(recall_score(eval_df['y_test'],eval_df['y_pred'],average='macro'))
print(recall_score(eval_df['y_test'],eval_df['y_pred'],average='weighted'))
```

```python
print(precision_score(eval_df['y_test'],eval_df['y_pred'],pos_label=0))
print(precision_score(eval_df['y_test'],eval_df['y_pred'],pos_label=1))
print(precision_score(eval_df['y_test'],eval_df['y_pred'],average='macro'))
print(precision_score(eval_df['y_test'],eval_df['y_pred'],average='weighted'))
```

```python
print(f1_score(eval_df['y_test'],eval_df['y_pred'],pos_label=0))
print(f1_score(eval_df['y_test'],eval_df['y_pred'],pos_label=1))
print(f1_score(eval_df['y_test'],eval_df['y_pred'],average='macro'))
print(f1_score(eval_df['y_test'],eval_df['y_pred'],average='weighted'))
```

```python
cr = classification_report(eval_df['y_test'],eval_df['y_pred'])
print(cr)
```
