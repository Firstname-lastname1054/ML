```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
from mlxtend.preprocessing import TransactionEncoder
from ipywidgets import interact
from ipywidgets import widgets
import os
os.chdir("E:/Machine_Learning/Datasets/")
```

#### read_csv file can only read lines of equal length but in dataset we dont have same number of elements in eah row 
#### so we have to do preprocessing to load the data 

```python
groceries = []
with open('Groceries.csv','r') as f:groceries = f.read()
groceries = groceries.split("\n")
groceries_list = []
for i in groceries:
    groceries_list.append(i.split(","))
```

## TransactionEncoder takes list of list as input,so we have converted data accordingly.

```python
te = TransactionEncoder()
te_arr = te.fit_transform(groceries_list)
te_arr
```

```python
fp_df = pd.DataFrame(te_arr,columns=te.columns_)
fp_df
```

```python
item_set = apriori(fp_df , min_support=  0.05 , use_colnames=True)
item_set
```

```python
rules = association_rules(item_set,metric = 'confidence',min_threshold = 0.2)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by =['support','confidence','lift'],ascending = False)
```

# Same process on 'DatasetA.csv'

```python
groceries = []
with open('DataSetA.csv','r') as f:groceries = f.read()
groceries = groceries.split("\n")
groceries_list = []
for i in groceries:
    groceries_list.append(i.split(","))
```

```python
te = TransactionEncoder()
te_arr = te.fit_transform(groceries_list)
fp_df = pd.DataFrame(te_arr,columns=te.columns_)
fp_df
```

### This dataset have traiiling comma at the end of every row thats why we are getting an extra column at first index as empty string ('') and we have to remove that

```python
fp_df = fp_df.iloc[:,1:]
fp_df.head()
```

```python
item_set = apriori(fp_df , min_support=  0.1 , use_colnames=True)
item_set
```

```python
rules = association_rules(item_set,metric = 'confidence',min_threshold = 0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by =['support','confidence','lift'],ascending = False)
```

```python
relevant_rules = rules[rules['lift']>1]
relevant_rules.sort_values(by = ['confidence','lift','support'],ascending=False)
```
