```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.ensemble import GradientBoostingClassifier,StackingClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.tree import DecisionTreeClassifier
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from tqdm import tqdm
import os
os.chdir('E:/Machine_Learning/Cases/Vehicle Silhouettes/')
```

```python
vehicle= pd.read_csv("Vehicle.csv")
vehicle.head()
```

```python
X,y = vehicle.drop('Class',axis=1),vehicle['Class']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
nb = GaussianNB()
lr = LogisticRegression()
lda = LinearDiscriminantAnalysis()
dtc = DecisionTreeClassifier(random_state = 26)
gbm = GradientBoostingClassifier(random_state=26)
```

```python
stack = StackingClassifier(estimators=[('TREE',dtc),('LR',lr),('LDA',lda),('NB',nb)],final_estimator=gbm)
stack.fit(X_train,y_train)
y_pred = stack.predict(X_test)
accuracy_score(y_test,y_pred)
```

### Using passthrough option

```python
stack = StackingClassifier(estimators=[('TREE',dtc),('LR',lr),('LDA',lda),('NB',nb)],final_estimator=gbm,passthrough=True)
stack.fit(X_train,y_train)
y_pred = stack.predict(X_test)
accuracy_score(y_test,y_pred)
```

## Using XGBoost(xgboost) instead of GradientBoosting (sklearn)

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test) 
```

```python
from xgboost import XGBClassifier
```

```python
xgb = XGBClassifier()
stack = StackingClassifier(estimators=[('TREE',dtc),('LR',lr),('LDA',lda),('NB',nb)],final_estimator=xgb)
stack.fit(X_train,y_train_encoded)
y_pred = stack.predict(X_test)
accuracy_score(y_test_encoded,y_pred)
```

#### With Passthrough

```python
xgb = XGBClassifier()
stack = StackingClassifier(estimators=[('TREE',dtc),('LR',lr),('LDA',lda),('NB',nb)],final_estimator=xgb,passthrough=True)
stack.fit(X_train,y_train_encoded)
y_pred = stack.predict(X_test)
accuracy_score(y_test_encoded,y_pred)
```
