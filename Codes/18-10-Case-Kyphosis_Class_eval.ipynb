```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_curve,roc_auc_score,RocCurveDisplay
from tqdm import tqdm
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
import os 
os.chdir('E:/Machine_Learning/Cases/Kyphosis/')
```

```python
df = pd.read_csv("Kyphosis.csv")
df
```

```python
X,y = df.drop('Kyphosis',axis=1),df['Kyphosis']
le = LabelEncoder()
le_y = le.fit_transform(y)
X_train,X_test,y_train,y_test = train_test_split(X,le_y,test_size=0.3,random_state=26,stratify=le_y)
```

```python
gbc = GradientBoostingClassifier(random_state=26)
gbc.fit(X_train,y_train)
y_pred = gbc.predict(X_test)
roc_auc_score(y_test,y_pred)
```

```python
gbc = GradientBoostingClassifier(random_state=26)
gbc.fit(X_train,y_train)
y_pred_prob = gbc.predict_proba(X_test)
roc_auc_score(y_test,y_pred_prob[:,1])
```

```python
learn_rates = np.linspace(0.01,0.9,10)
max_depths = [None,2,3,4]
n_estimators = [25,50,100]
scores = []
for l in tqdm(learn_rates):
    for m in max_depths:
        for n in n_estimators:
            gbm = GradientBoostingClassifier(random_state=26,learning_rate=l,max_depth=m,n_estimators=n)
            gbm.fit(X_train,y_train)
            y_pred = gbm.predict_proba(X_test)
            scores.append([l,m,n,roc_auc_score(y_test,y_pred_prob[:,1])])
df_scores = pd.DataFrame(scores,columns=['Learning_rate','Max_depth','n_estimators','roc_auc_score'])
df_scores.sort_values(by='roc_auc_score',ascending=False)
```
